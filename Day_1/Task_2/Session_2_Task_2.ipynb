Task 1: Tokenization (Word & Sentence Level)

Goal: Understand how to split raw text into clean tokens at both sentence and word levels.

```python
import nltk
# nltk.download('punkt')
# nltk.download('punkt_tab')
from nltk.tokenize import word_tokenize, sent_tokenize

text = "Hello students! Welcome to NLP Session 2. We're learning preprocessing today, aren't we?"

# 1. Split into sentences
sentences = nltk.sent_tokenize(text)

print("Sentence Tokenizer:")
print(sentences)

print()

# 2. Split into words
words = nltk.word_tokenize(text)

print("Word Tokenizer:")
print(words)
```

Task 2: Regular Expressions (re Cleaning)

Goal: Clean noise (URLs, special characters, numbers, and extra spaces) from messy text before passing it to NLP models.

```python
import re

raw_tweet = "Great session today!! Check out http://example.com for NLP code @student123 #AI2026"

# print each one alone
URLs = re.findall(r'https?://[^\s]+', raw_tweet)
print(URLs)

special_Characters = re.findall(r'[^a-zA-Z0-9\s]', raw_tweet)
print(special_Characters)

numbers = re.findall(r'\d+', raw_tweet)
print(numbers)

extra_Spaces = re.findall(r'\s{2,}', raw_tweet)
print(extra_Spaces)

print()

# add all of them to a list
items_to_remove = URLs + special_Characters + extra_Spaces + numbers

# remove them from the text
for item in items_to_remove:
    raw_tweet = raw_tweet.replace(item, "")

raw_tweet = " ".join(raw_tweet.split())
print(raw_tweet)
```

Task 3: Stop Words Removal

Goal: Filter out frequent, low-information words ("the", "is", "in") to focus on content-bearing tokens.

```python
import nltk
# nltk.download('stopwords')
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

# call stop words dictionary
stop_words = set(stopwords.words('english'))

text = "This is a simple demo showing how stop words are filtered out from a sentence."

# split words
words = word_tokenize(text)

# remove stop words from the text
filtered_text = [word for word in words if word.lower() not in stop_words]
print(filtered_text)

print()

filtered_text = " ".join(filtered_text)
print(filtered_text)
```

Task 4: Stemming vs. Lemmatization

Goal: See the difference between crude suffix stripping (Stemming) and dictionary/grammar-based root extraction (Lemmatization).

```python
import nltk
# nltk.download('wordnet')
# nltk.download('omw-1.4')
from nltk.stem import PorterStemmer, WordNetLemmatizer

words = ["playing", "plays", "played", "easily", "careful"]

stemmer = PorterStemmer()
lemmatizer = WordNetLemmatizer()

stemmed_words = [stemmer.stem(word) for word in words]
lemmatized_words = [lemmatizer.lemmatize(word) for word in words]

# stemmer cuts the text with what he sees right
# but lemmatizer returns the words to its origin from the dictionary itself not randomly
print("Stemmed Words:   ", stemmed_words)
print("Lemmatized Words:", lemmatized_words)
```
