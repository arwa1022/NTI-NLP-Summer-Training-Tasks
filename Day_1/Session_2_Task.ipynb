#Exercise: "Build a Toy Sentiment Classifier"

Task: Students use positive and negative word lists to score a review.



```python
# Define sentiment vocabularies
positive_words = ["love", "great", "awesome", "fast", "good"]
negative_words = ["bad", "slow", "terrible", "hate", "worst"]

# Input sentence
# review = "The camera is great and awesome but the battery is bad"

# review = """
# I love this product! The quality is great, and the shipping was fast.
# Everything worked as expected, and the overall experience was good.
# It's an awesome purchase, and I would definitely recommend it to others."""

review = """
I hate this product. The delivery was slow, and the quality was bad.
It wasn't completely unusable, but the experience was disappointing. It's far from the worst thing I've bought,
but I definitely expected better."""

positive_words_count = 0
negative_words_count = 0

words = review.lower().split()

for word in words:
  if word in positive_words:
    positive_words_count += 1
  elif word in negative_words:
    negative_words_count += 1

print("positive words count:" + str(positive_words_count))
print("negative words count:" + str(negative_words_count))
print()

if positive_words_count > negative_words_count:
  print("positive review")
elif negative_words_count > positive_words_count:
  print("negative review")
else:
  print("neutral review")
```

#Example 2: Application (Spam Filter / Text Classification)

Concept: A basic rule-based application demonstrating how NLP systems classify text using predefined vocabulary checks.

```python
# Define vocabulary rules for spam detection
spam_keywords = ["free", "win", "cash", "click", "prize", "urgent"]

# email_text = """
# Dear Customer,

# This is URGENT! You have been selected to WIN a FREE CASH PRIZE worth $5,000.
# CLICK the link below immediately to claim your reward before it expires. This exclusive offer is available for a limited time only.
# Don't miss your chance to WIN FREE CASH. CLICK now to receive your PRIZE!

# Best regards,
# Rewards Team
# """

email_text = """
Hello Sarah,

I hope you're doing well.
Just a reminder that our project meeting is scheduled for tomorrow at 10:00 AM. Please review the attached document before the meeting.
If you'd like, you can click the shared link to access the latest version of the presentation.
Looking forward to discussing the project with everyone.

Best regards,
Michael Johnson
"""

def classify_email(email_text):
    count = 0
    words = email_text.lower().split()
    for i in words:
      if i in spam_keywords:
        count += 1
    return count

print(classify_email(email_text))
print()

if classify_email(email_text) > 3:
  print("Spam")
else:
  print("not spam")
```
